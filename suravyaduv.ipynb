```python
def extract_likers_from_popup(likers_popup):
    import sqlite3
    from selenium.webdriver.common.by import By

    likers_data = []

    try:
        # Find all profile links
        profile_links = likers_popup.find_elements(
            By.XPATH, ".//a[starts-with(@href, 'https://www.facebook.com/') and @role='link' and @tabindex='0' and not(contains(@href, 'privacy'))]"
        )
        print(f"[DEBUG] Found {len(profile_links)} profile links in popup")

        for i, link in enumerate(profile_links):
            try:
                name = link.text.strip()
                href = link.get_attribute("href")

                if "profile.php" in href and "id=" in href:
                    href = href.split("&")[0]

                if name and len(name) > 2:
                    likers_data.append((name, href))
                    print(f"[DEBUG] Found liker: {name} -> {href}")
                else:
                    print(f"[DEBUG] Skipped link {i+1}: Invalid name '{name}'")

            except Exception as e:
                print(f"[ERROR] Failed to process link {i+1}: {e}")

        # Remove duplicates within the same popup
        unique_likers = []
        seen = set()
        for name, href in likers_data:
            if href not in seen:
                seen.add(href)
                unique_likers.append((name, href))

        # ===============================
        # Save to SQLite Database
        # ===============================
        if unique_likers:
            conn = sqlite3.connect("facebook_likers.db")  # Create DB if it doesn't exist
            cursor = conn.cursor()

            # Create table if not exists
            cursor.execute('''
                CREATE TABLE IF NOT EXISTS likers (
                    id INTEGER PRIMARY KEY AUTOINCREMENT,
                    name TEXT NOT NULL,
                    profile_link TEXT UNIQUE
                )
            ''')

            inserted_count = 0
            for name, link in unique_likers:
                try:
                    cursor.execute("INSERT OR IGNORE INTO likers (name, profile_link) VALUES (?, ?)", (name, link))
                    if cursor.rowcount:
                        inserted_count += 1
                except Exception as db_err:
                    print(f"[ERROR] DB insert error: {db_err}")

            conn.commit()
            conn.close()

            print(f"[INFO] Inserted {inserted_count} new likers into SQLite database.")
        else:
            print("[INFO] No likers found.")

    except Exception as e:
        print(f"[ERROR] Failed to extract likers: {e}")
```

```python
# This Code 

def scrape_likers_from_all_posts(driver, output_file="likers.csv", max_posts=10):
    from selenium.webdriver.common.by import By
    from selenium.webdriver.support.ui import WebDriverWait
    from selenium.webdriver.support import expected_conditions as EC
    import time, csv, os

    print("[INFO] Starting Facebook liker scraper with smooth scrolling...")

    # Step 1: Click the 'Posts' tab
    try:
        posts_tab = WebDriverWait(driver, 10).until(
            EC.element_to_be_clickable((By.XPATH, "//span[text()='Posts']"))
        )
        driver.execute_script("arguments[0].click();", posts_tab)
        print("[DEBUG] Clicked Posts tab.")
        time.sleep(3)
    except Exception as e:
        print("[ERROR] Could not find Posts tab:", e)
        return

    # Step 2: Load all posts using smooth scrolling
    print("[INFO] Smooth scrolling to load all posts...")
    post_xpath = "//div[contains(@class, 'x1yztbdb') and contains(@class, 'x1n2onr6') and contains(@class, 'xh8yej3') and contains(@class, 'x1ja2u2z')]"
    posts = driver.find_elements(By.XPATH, post_xpath)
    last_count = 0
    max_attempts = 30
    attempts = 0
    
    while len(posts) > last_count and attempts < max_attempts:
        last_count = len(posts)
        
        # Use smooth scroll for posts
        for i in range(3):  # Multiple smooth scrolls
            smooth_scroll(driver, 800)  # Scroll down by 800px smoothly
        
        time.sleep(3)  # Wait for content to load
        posts = driver.find_elements(By.XPATH, post_xpath)
        print(f"[DEBUG] After smooth scrolling, found {len(posts)} posts")
        attempts += 1

    print(f"[INFO] Total posts collected after smooth scrolling: {len(posts)}")

    # Step 3: Process each post
    for post_idx in range(min(len(posts), max_posts)):
        try:
            print(f"\n[STEP] Processing post {post_idx + 1}")
            posts = driver.find_elements(By.XPATH, post_xpath)
            if post_idx >= len(posts):
                print(f"[WARN] Post index {post_idx+1} out of range after scrolling. Skipping.")
                continue
            post_elem = posts[post_idx]
            
            # Smooth scroll to the post
            driver.execute_script("arguments[0].scrollIntoView({block: 'center'});", post_elem)
            time.sleep(3)

            # Find the like button only within this post
            like_btn = None
            try:
                like_buttons = post_elem.find_elements(
                    By.XPATH,
                    ".//div[@role='button' and .//div[contains(text(), 'All reactions')]]"
                )
                print(f"[DEBUG] Found {len(like_buttons)} 'All reactions' buttons in post {post_idx+1}")
                for btn in like_buttons:
                    if btn.is_displayed() and btn.is_enabled():
                        like_btn = btn
                        break
            except Exception as e:
                print(f"[WARN] Error finding like button for post {post_idx+1}: {e}")
                continue
                
            if not like_btn:
                print(f"[WARN] No visible/clickable like button for post {post_idx+1}, skipping.")
                continue
                
            try:
                driver.execute_script("arguments[0].scrollIntoView(true);", like_btn)
                like_btn.click()
                print("[DEBUG] Clicked 'All reactions' button.")
                time.sleep(3)
            except Exception:
                try:
                    driver.execute_script("arguments[0].click();", like_btn)
                    print("[DEBUG] JS clicked 'All reactions' button.")
                    time.sleep(3)
                except Exception:
                    print(f"[WARN] Could not click like button for post {post_idx+1}, skipping.")
                    continue

            # Step 4: Wait for popup window
            try:
                likers_popup = WebDriverWait(driver, 15).until(
                    EC.presence_of_element_located((By.XPATH, "//div[@role='dialog' and not(contains(@aria-label, 'Messenger')) and not(contains(@aria-label, 'Notifications'))]"))
                )
                print("[DEBUG] Reactions popup appeared.")
                time.sleep(5)  # Wait for popup to load
            except Exception as e:
                print("[ERROR] Popup did not appear:", e)
                continue

            # Step 5: Smooth scroll popup to load all likers
            print("[DEBUG] Starting smooth scrolling to load all likers...")
            
            # Get initial liker count
            initial_profile_links = likers_popup.find_elements(By.XPATH, ".//a[starts-with(@href, 'https://www.facebook.com/') and @role='link' and @tabindex='0' and not(contains(@href, 'privacy'))]")
            initial_liker_count = len(initial_profile_links)
            print(f"[DEBUG] Initial liker count: {initial_liker_count}")
            
            # Find the scrollable container within the popup
            scroll_box = None
            
            # Look for scrollable containers
            scroll_containers = likers_popup.find_elements(By.XPATH, ".//div[contains(@class, 'x1n2onr6') and contains(@class, 'x1ja2u2z')]")
            for container in scroll_containers:
                try:
                    scroll_height = driver.execute_script("return arguments[0].scrollHeight", container)
                    client_height = driver.execute_script("return arguments[0].clientHeight", container)
                    if scroll_height > client_height + 10:
                        scroll_box = container
                        print(f"[DEBUG] Found scrollable container with height {scroll_height} > {client_height}")
                        break
                except:
                    continue
            
            # If no scrollable container found, try the popup itself
            if not scroll_box:
                try:
                    scroll_height = driver.execute_script("return arguments[0].scrollHeight", likers_popup)
                    client_height = driver.execute_script("return arguments[0].clientHeight", likers_popup)
                    if scroll_height > client_height + 10:
                        scroll_box = likers_popup
                        print(f"[DEBUG] Using popup itself as scrollable container")
                except:
                    print("[DEBUG] Could not determine scrollable content")
            
            # Smooth scroll the popup to load all content
            if scroll_box:
                print("[DEBUG] Starting smooth scrolling to load all likers...")
                
                scroll_attempts = 0
                max_scroll_attempts = 50  # High for smooth scrolling
                last_height = driver.execute_script("return arguments[0].scrollHeight", scroll_box)
                no_change_count = 0
                last_link_count = initial_liker_count
                
                print(f"[DEBUG] Starting smooth scroll. Initial height: {last_height}, Initial links: {initial_liker_count}")
                
                while scroll_attempts < max_scroll_attempts:
                    try:
                        # Multiple smooth scrolling strategies
                        # Strategy 1: Smooth scroll to bottom
                        driver.execute_script(f"arguments[0].scrollTop += 500;", scroll_box)
                        time.sleep(4)
                        
                        # Strategy 2: Smooth incremental scrolling
                        if scroll_attempts % 3 == 0:
                            driver.execute_script(f"arguments[0].scrollTop += 800;", scroll_box)
                            time.sleep(4)
                        
                        # Strategy 3: Smooth percentage-based scrolling
                        if scroll_attempts % 5 == 0:
                            scroll_height = driver.execute_script("return arguments[0].scrollHeight", scroll_box)
                            target_scroll = scroll_height * 0.7
                            current_scroll = driver.execute_script("return arguments[0].scrollTop", scroll_box)
                            driver.execute_script(f"arguments[0].scrollTop = {target_scroll};", scroll_box)
                            time.sleep(4)
                        
                        # Strategy 4: Smooth scroll to very bottom
                        if scroll_attempts % 7 == 0:
                            scroll_height = driver.execute_script("return arguments[0].scrollHeight", scroll_box)
                            driver.execute_script(f"arguments[0].scrollTop = {scroll_height};", scroll_box)
                            time.sleep(4)
                        
                        # Check if new content loaded
                        new_height = driver.execute_script("return arguments[0].scrollHeight", scroll_box)
                        profile_links = likers_popup.find_elements(By.XPATH, ".//a[starts-with(@href, 'https://www.facebook.com/') and @role='link' and @tabindex='0' and not(contains(@href, 'privacy'))]")
                        current_link_count = len(profile_links)
                        
                        print(f"[DEBUG] Smooth scroll attempt {scroll_attempts + 1}: Height {last_height} -> {new_height}, Links {last_link_count} -> {current_link_count}")
                        
                        if new_height == last_height and current_link_count == last_link_count:
                            no_change_count += 1
                            if no_change_count >= 8:  # Higher tolerance for smooth scrolling
                                print(f"[DEBUG] No more content to load after {no_change_count} attempts. All likers loaded.")
                                break
                        else:
                            no_change_count = 0  # Reset counter if new content found
                            last_height = new_height
                            last_link_count = current_link_count
                            print(f"[DEBUG] New content loaded. Height: {last_height}, Links: {current_link_count}")
                        
                        scroll_attempts += 1
                        
                    except Exception as e:
                        print(f"[DEBUG] Error during smooth scrolling: {e}")
                        break
                
                print(f"[DEBUG] Finished smooth scrolling after {scroll_attempts} attempts. Final height: {last_height}")
                print(f"[DEBUG] Total likers loaded: {current_link_count}")
            else:
                print("[DEBUG] No scrollable content found, proceeding with extraction")
            
            # Wait for all content to settle
            time.sleep(5)
            print("[DEBUG] All likers loaded. Now extracting...")

            # Step 6: Extract liker data
            try:
                extract_likers_from_popup(likers_popup)
                print(f"[INFO] Extracted likers for post {post_idx + 1}")
            except Exception as e:
                print(f"[ERROR] Failed to extract likers: {e}")

            # Step 7: Close the popup
            try:
                close_buttons = likers_popup.find_elements(By.XPATH, ".//div[@role='button' and @aria-label='Close']")
                if close_buttons:
                    driver.execute_script("arguments[0].click();", close_buttons[0])
                    print("[DEBUG] Closed popup using close button.")
                    time.sleep(4)
                else:
                    from selenium.webdriver.common.keys import Keys
                    driver.find_element(By.TAG_NAME, "body").send_keys(Keys.ESCAPE)
                    print("[DEBUG] Closed popup using Escape key.")
                    time.sleep(4)
            except Exception as e:
                print(f"[WARN] Could not close popup: {e}")
                try:
                    from selenium.webdriver.common.keys import Keys
                    driver.find_element(By.TAG_NAME, "body").send_keys(Keys.ESCAPE)
                    print("[DEBUG] Closed popup using Escape key (fallback).")
                    time.sleep(4)
                except:
                    pass

        except Exception as e:
            print(f"[ERROR] Error processing post {post_idx + 1}: {e}")
            time.sleep(3)
```
